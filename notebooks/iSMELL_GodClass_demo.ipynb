## 2. Load the curated detection dataset
The replication package bundles expert-tool outputs in `CodeDetection/updated_dataset.xlsx`.

````python
dataset_path = Path('../CodeDetection/updated_dataset.xlsx')
if not dataset_path.exists():
    raise FileNotFoundError(f'Missing data file: {dataset_path}')
df = pd.read_excel(dataset_path)
df.head()
````

## 3. Focus on God Class samples
Filter the entries that correspond to God Class smells and inspect a subset with the raw tool decisions.

````python
tools = ['PMD','Jdeodorant','organic','DesigniteJava','JSpIRIT','FeTruth','Jmove']
god_df = df[df['Smell'] == 'GodClass'].copy()
god_df['Span'] = (god_df['Starting Line Number'].fillna(0).astype(int).astype(str) +
                 ' - ' + god_df['Ending Line Number'].fillna(0).astype(int).astype(str))
columns = ['Project','Class','Span','Link'] + tools + ['Results']
god_df[columns].head(10)
````

## 4. Summarize detections by expert tool
Count how frequently each external detector flags the God Class smell.

````python
summary = (god_df[tools]
           .applymap(lambda x: 1 if x == 'Detected' else 0)
           .sum()
           .rename('DetectedCount'))
summary.to_frame().T
````

## 5. Agreement patterns across tools
Review the most common detection combinations to see how tools agree or disagree.

````python
pattern_counts = (god_df[tools]
                  .apply(lambda row: tuple(row.values), axis=1)
                  .value_counts()
                  .head(5))
pattern_counts
````
